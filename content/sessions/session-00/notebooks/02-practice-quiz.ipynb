## 1. Tokenisation

Is one English word always one token? Explain what the vocabulary stores.

<details><summary>Show answer</summary>

No. A tokenizer may represent a word using one or several tokens, and may also tokenize punctuation, whitespace or bytes. Its vocabulary maps each supported token unit to an integer ID. The ID selects an embedding-table row; it is not a semantic measurement.

</details>

Pretraining primarily teaches sequence continuation. Instruction/chat behavior comes from post-training and chat conventions that reward following user intent.

</details>

## 5. Cosine similarity

Without calculating, predict the cosine similarity of `[1, 1]` with `[10, 10]`, and with `[1, -1]`.

<details><summary>Show answer</summary>

The first is `1`: same direction despite different magnitude. The second is `0`: the vectors are orthogonal. Cosine similarity compares direction after accounting for magnitude.

</details>

```python
# Check your prediction without a library.
from math import sqrt


def cosine(a, b):
    dot = sum(x * y for x, y in zip(a, b, strict=True))
    return dot / (sqrt(sum(x*x for x in a)) * sqrt(sum(y*y for y in b)))

cosine([1, 1], [10, 10]), cosine([1, 1], [1, -1])
```

## 6. Architecture choice

An application must answer questions about a private handbook updated weekly and cite its sources. Choose direct LLM, RAG, or an agent.

<details><summary>Show answer</summary>

Start with RAG: retrieve current handbook passages, place them in context and return citations. An agent is unnecessary unless retrieval requires adaptive tool choice or multi-step investigation.

</details>

## 7. Workflow versus agent

A refund always requires validation, fraud screening, approval above a threshold, then payment. Would you begin with an agent?

<details><summary>Show answer</summary>

No. The path is known, so a deterministic workflow is easier to test, audit and operate. An LLM might help classify evidence inside a bounded step without controlling the workflow.

</details>

## 8. Loop and harness

Classify each concern: choosing the next tool, a maximum of five steps, tool permissions, deciding the goal is complete, and recording traces.

<details><summary>Show answer</summary>

Choosing the next tool and deciding completion are loop behavior. Step limits, permissions and tracing are harness controls. In real systems the boundary can overlap, but the distinction reveals who controls risk.

</details>
